```python
# intv100

maxIter   : int   = 500
intv      : int   = 100
basis     : float = 0.01
start     : int   = 1
end       : int   = 10000
group     : str   = "type 3 18"
indole_idx: int   = 1
prefix    : str   = 'beta'

benchmark_params = [
    f'data/benchmark_nve_basis{basis}_intv{intv}/' if prefix == '' else f'data/{prefix}_benchmark_nve_basis{basis}_intv{intv}/',
    f'frames{start}_{end}.npz', 
    f'GlobalVariable.npz',
    f'rdf_indole2zeo.dat',
]
control_params = [
    f'data/control_nve_basis{basis}_intv{intv}/' if prefix == '' else f'data/{prefix}_control_nve_basis{basis}_intv{intv}/',
    f'frames{start}_{end}.npz', 
    f'GlobalVariable.npz',
    f'rdf_indole2zeo.dat',
]
edsr_params = [
    f'data/EdSr_nve_basis{basis}_intv{intv}_iter500/' if prefix == '' else f'data/{prefix}_EdSr_nve_basis{basis}_intv{intv}_iter500/',
    f'frames{start}_{end}.npz', 
    f'GlobalVariable.npz',
    f'rdf_indole2zeo.dat',
]

# mass shape: (natoms,), id shape: (natoms,),  x shape: (ntrajs, natoms, 3), v shape: (ntrajs, natoms, 3)
with ThreadPoolExecutor(max_workers = 3) as executor:
    futures = [
        executor.submit(extraction, *benchmark_params[:-1], filter = group),
        executor.submit(extraction, *control_params[:-1], filter = group),
        executor.submit(extraction, *edsr_params[:-1] , filter = group),
    ]

benchmark_x, benchmark_v, benchmark_delta_t, benchmark_mass, benchmark_atype, benchmark_id, benchmark_boundary, benchmark_heads, benchmark_ppties, benchmark_init_state, benchmark_last_state = futures[0].result()
control_x, control_v, control_delta_t, control_mass, control_atype, control_id, control_boundary, control_heads, control_ppties, control_init_state, control_last_state = futures[1].result()
edsr_x, edsr_v, edsr_delta_t, edsr_mass, edsr_atype, edsr_id, edsr_boundary, edsr_heads, edsr_ppties, edsr_init_state, edsr_last_state = futures[2].result()


print('before sort:')
print(benchmark_id, benchmark_atype)
print(control_id, control_atype)
print(edsr_id, edsr_atype)


bid_group = np.argsort(benchmark_id)
cid_group = np.argsort(control_id)
tid_group = np.argsort(edsr_id)



benchmark_x, benchmark_v, benchmark_mass, benchmark_atype, benchmark_id = \
    benchmark_x[:, bid_group], benchmark_v[:, bid_group], benchmark_mass[bid_group], benchmark_atype[bid_group], benchmark_id[bid_group]

control_x, control_v, control_mass, control_atype, control_id = \
    control_x[:, cid_group], control_v[:, cid_group], control_mass[cid_group], control_atype[cid_group], control_id[cid_group]

edsr_x, edsr_v, edsr_mass, edsr_atype, edsr_id = \
    edsr_x[:, tid_group], edsr_v[:, tid_group], edsr_mass[tid_group], edsr_atype[tid_group], edsr_id[tid_group]

print('after sort:')
print(benchmark_id, benchmark_atype)
print(control_id, control_atype)
print(edsr_id, edsr_atype)


benchmark_x, benchmark_v, benchmark_mass, benchmark_atype, benchmark_id = \
    benchmark_x[:,indole_idx * 16:(indole_idx + 1) *16], benchmark_v[:, indole_idx * 16:(indole_idx + 1) *16], benchmark_mass[indole_idx * 16:(indole_idx + 1) *16], benchmark_atype[indole_idx * 16:(indole_idx + 1) *16], benchmark_id[indole_idx * 16:(indole_idx + 1) *16]

control_x, control_v, control_mass, control_atype, control_id = \
    control_x[:,indole_idx * 16:(indole_idx + 1) *16], control_v[:, indole_idx * 16:(indole_idx + 1) *16], control_mass[indole_idx * 16:(indole_idx + 1) *16], control_atype[indole_idx * 16:(indole_idx + 1) *16], control_id[indole_idx * 16:(indole_idx + 1) *16]

edsr_x, edsr_v, edsr_mass, edsr_atype, edsr_id = \
    edsr_x[:,indole_idx * 16:(indole_idx + 1) *16], edsr_v[:, indole_idx * 16:(indole_idx + 1) *16], edsr_mass[indole_idx * 16:(indole_idx + 1) *16], edsr_atype[indole_idx * 16:(indole_idx + 1) *16], edsr_id[indole_idx * 16:(indole_idx + 1) *16]

print('\n\nafter filter:')
print(benchmark_id, benchmark_atype)
print(control_id, control_atype)
print(edsr_id, edsr_atype)

times = np.arange(end)*intv*basis
```

```python
def read(path, file) -> np.ndarray:
    with open(path + '/' + file, 'r') as f:
        data = f.read().replace('\n', ' ').split(' ')
        if data[-1] == '':
            data.pop(-1)

    data = list(map(float, data))
    data = rearrange(np.array(data), "(npoints ndims) -> ndims npoints", ndims = 3)
    
    return data

benchmark_rdf = read(benchmark_params[0], benchmark_params[-1])
control_rdf = read(control_params[0], control_params[-1])
edsr_rdf = read(edsr_params[0], edsr_params[-1])

absve_bc = np.fabs(benchmark_rdf[1] - control_rdf[1])
absve_bt = np.fabs(benchmark_rdf[1] - edsr_rdf[1])

# ! Picture idx
pic_idx += 1
subgraph_item += 1

ax = plt.subplot(row, col, 1)
plt.text(*item_pos, chr(subgraph_item), transform = ax.transAxes, fontsize = item_fontsize.get_size(), fontweight = 'bold', va = 'top', ha = 'left')

# ! title of column 
plt.text(0.5, 1.15, "Radial distribution function", transform = ax.transAxes, fontsize = title_fontsize.get_size(), fontweight = 'bold', va = 'center', ha = 'center')

# ! title of row
plt.text(-0.2, 0.5, f"{basis*intv:.1f} fs", transform = ax.transAxes, fontsize = title_fontsize.get_size(), fontweight = 'bold', va = 'center', ha = 'center')
plt.plot(benchmark_rdf[0], benchmark_rdf[1], label = f"{benchmark_name}", linewidth = 4)
plt.plot(edsr_rdf[0], edsr_rdf[1], label = f"{model_name}", linewidth = 2)
plt.plot(control_rdf[0], control_rdf[1], label = f"{control_name}", linewidth = 1)

if tight_layout_arg is not None:
    plt.subplots_adjust(**tight_layout_arg)
plt.legend(fontsize = legned_fontsize.get_size(), loc = 'lower right', ncol = 1)
```

```python
# velocity distribution of all atom
bix, biv, bitype, biid, bim  = benchmark_init_state
blx, blv, bltype, blid, blm  = benchmark_last_state

cix, civ, citype, ciid, cim  = control_init_state
clx, clv, cltype, clid, clm  = control_last_state

tix, tiv, titype, tiid, tim  = edsr_init_state
tlx, tlv, tltype, tlid, tlm  = edsr_last_state

type, st, en = group.split(" ")
if type == 'id':
    mask = np.logical_and(biid >= int(st), biid <= int(en))
elif type == 'type':
    mask = np.logical_and(bitype >= int(st), bitype <= int(en))

    
blv_norm = np.sqrt(np.sum(blv**2, axis = -1))
tlv_norm = np.sqrt(np.sum(tlv**2, axis = -1))
clv_norm = np.sqrt(np.sum(clv**2, axis = -1))

axis_left, axis_right = min(blv.min(), tlv.min(), clv.min()), max(blv.max(), tlv.max(), clv.max())

# ! Picture idx
pic_idx += 1
subgraph_item += 1

ax = plt.subplot(row, col, pic_idx)
print(ax)
plt.text(*item_pos, chr(subgraph_item), transform = ax.transAxes, fontsize = fontsize.get_size(), fontweight = 'bold', va = 'top', ha = 'left')

# ! title of column 
plt.text(0.5, 1.15, "Velocity distribution", transform = ax.transAxes, fontsize = title_fontsize.get_size(), fontweight = 'bold', va = 'center', ha = 'center')

plt.hist([blv[mask].reshape(-1), tlv[mask].reshape(-1)], bins = 20, density = True, histtype = 'bar', range = (axis_left, axis_right), edgecolor='white', color = ['b', 'orange'], align = 'left', label = [f'{benchmark_name}', f'{model_name}'], rwidth = 1.0)
plt.tick_params(axis = 'y', labelsize = tick_fontsize.get_size() - 1)
plt.xticks([-0.06, -0.04, -0.02, 0, 0.02, 0.04, 0.06]); plt.xlim(right = 0.11)
plt.yticks([0, 10, 20, 30, 40]); plt.ylim(top = 42)
if tight_layout_arg is not None:
    plt.subplots_adjust(**tight_layout_arg)
plt.legend(fontsize = legned_fontsize.get_size(), loc = 'upper left', ncol = 1)

box = ax.get_position()
x0, y0, x1, y1 = box.x0, box.y0, box.x1, box.y1

aia = fig.add_axes([x0 + 0.21, y0 + 0.15, 0.2, 0.2])
aia.hist([blv[mask].reshape(-1), clv[mask].reshape(-1)], bins = 20, density = True, histtype = 'bar', range = (axis_left, axis_right), edgecolor='white', color = ['b', 'r'], align = 'mid', label = [f'{benchmark_name}', f'{control_name}'], rwidth = 1.0)
aia.tick_params(axis = 'both', labelsize = 7)
aia.set_xticks([-0.06, -0.04, -0.02, 0, 0.02, 0.04, 0.06])
aia.set_yticks([0, 10, 20, 30, 40]); aia.set_ylim(top = 42)
aia.legend(fontsize = fontsize.get_size() - 3, loc = 'upper right', ncol = 1)
```
